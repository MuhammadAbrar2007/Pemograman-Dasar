```python
# accumulator
scores = [80, 65, 90, 75, 70]

total = 0
count_pass = 0
for score in scores:
    total = total + score
    if score >= 75:
        count_pass = count_pass + 1
    print("running total:", total)

average = total / len(scores)
print("Total   :", total)
print("Average :", average)
print("Scores >= 75:", count_pass, "of", len(scores))
```

```python
# highest lowest
scores = [80, 65, 90, 75, 70]

highest = scores[0]
lowest = scores[0]
for score in scores:
    if score > highest:
        highest = score
    if score < lowest:
        lowest = score

print("Highest:", highest)
print("Lowest :", lowest)
print("Built-in check:", max(scores), min(scores))
```

```python
# filtering
temps = [28, 31, 30, 33, 29]

hot = []
for t in temps:
    if t >= 30:
        hot.append(t)
print("Hot values  :", hot)
print("How many    :", len(hot))

hot_days = []
for i in range(len(temps)):
    if temps[i] >= 30:
        hot_days.append(i + 1)
print("Hot days    :", hot_days)

hot_again = [t for t in temps if t >= 30]
print("Same result?", hot_again == hot)
```

```python
# traversal
energy = [
    5576, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083,
    6097, 6089, 6066, 5533, 5566, 6097, 3592, 6059, 6100, 6098,
    5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062,
]

total = 0
i_max = 0
i_min = 0
for i in range(len(energy)):
    total = total + energy[i]
    if energy[i] > energy[i_max]:
        i_max = i
    if energy[i] < energy[i_min]:
        i_min = i

print(f"Energy this month: {total:,} MWh")
print(f"Average          : {total / len(energy):,.0f} MWh/day")
print(f"Highest          : day {i_max + 1} ({energy[i_max]:,} MWh)")
print(f"Lowest           : day {i_min + 1} ({energy[i_min]:,} MWh)")
print("Built-ins        :", sum(energy), max(energy), min(energy))
```

```python
# search filter
energy = [
    5576, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083,
    6097, 6089, 6066, 5533, 5566, 6097, 3592, 6059, 6100, 6098,
    5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062,
]
target = 5800

first_day = None
for i in range(len(energy)):
    if energy[i] < target:
        first_day = i + 1
        break
print("First day below target:", first_day)


short_days = []
for i in range(len(energy)):
    if energy[i] < target:
        short_days.append(i + 1)
print("Days below target:", short_days)


short_days_2 = [i + 1 for i in range(len(energy)) if energy[i] < target]
print("Same result?", short_days == short_days_2)
print(f"Target met on {len(energy) - len(short_days)} of {len(energy)} days")
```

```python
# sequential
energy = [
    5576, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083,
    6097, 6089, 6066, 5533, 5566, 6097, 3592, 6059, 6100, 6098,
    5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062,
]

for i in range(1, len(energy)):
    change = (energy[i] - energy[i - 1]) / energy[i - 1] * 100
    if change < -20:
        print(f"Day {i + 1}: {change:+.1f}% versus the day before, investigate")

avg_7 = []
for i in range(6, len(energy)):
    window = energy[i - 6 : i + 1]
    avg_7.append(sum(window) / 7)

print(f"{len(avg_7)} moving-average values ({len(energy)} days of data)")
print(f"Lowest {min(avg_7):,.0f} MWh/day, highest {max(avg_7):,.0f} MWh/day")
```

```python
# weekday weekend
energy = [
    5576, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083,
    6097, 6089, 6066, 5533, 5566, 6097, 3592, 6059, 6100, 6098,
    5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062,
]
days = list(range(1, 31))
run_hours = [24] * 30
run_hours[16] = 18
DAY_NAMES = ["Sun", "Mon", "Tue", "Wed", "Thu", "Fri", "Sat"]

for i in range(len(energy)):
    if run_hours[i] < 24:
        print("Day", days[i], ":", run_hours[i], "h,", energy[i], "MWh")

weekday = []
weekend = []
for i in range(len(energy)):
    if DAY_NAMES[i % 7] == "Sat" or DAY_NAMES[i % 7] == "Sun":
        weekend.append(energy[i])
    else:
        weekday.append(energy[i])

print("Weekday average:", round(sum(weekday) / len(weekday)), "MWh")
print("Weekend average:", round(sum(weekend) / len(weekend)), "MWh")
```
