```python
#Nama : Muhammad Abrar Al Aziz
#NIM : 4232511044
#Tanggal : 13-11-26
#Mata Kuliah : Pemrograman Dasar
#Soal 9

data = ["1.2", "0.9", "ERR", "3.1", "1.8"]
data_tidak_valid = 0

for n, teks in enumerate(data, start=1):
    try:
        nilai = float(teks)
    except ValueError:
        print(f"Unit {n}: data tidak valid")
        data_tidak_valid += 1
        continue

    if nilai > 2.0:
        print(f"Warning: Unit {n} vibrasi tinggi!")
        if nilai > 3.0:
            print("KRITIS: pemeriksaan dihentikan")
            break

print(f"Jumlah data tidak valid: {data_tidak_valid}")
```
