```python
tegangan_harian = [217.5, 219.0, 220.3, 218.7, 221.1, 216.9]

# (a) Cetak pembacaan pertama dan terakhir
print(f"Pembacaan pertama: {tegangan_harian[0]} V")
print(f"Pembacaan terakhir: {tegangan_harian[-1]} V")

# (b) Tambahkan satu pembacaan baru 222.0 dengan append()
tegangan_harian.append(222.0)
print(f"\nDaftar tegangan setelah penambahan: {tegangan_harian}")

# (c) Cetak jumlah total data dengan len()
print(f"Jumlah total data: {len(tegangan_harian)}")
```

```python
# Buat dictionary spesifikasi motor listrik
spek_motor = {
    "nama": "Motor Induksi 3 Fasa",
    "daya_kW": 7.5,
    "tegangan_V": 400,
    "rpm": 1450
}

print("Spesifikasi motor awal:")
print(spek_motor)

# Cetak salah satu nilai lewat key-nya (contoh: daya_kW)
print(f"\nDaya motor: {spek_motor['daya_kW']} kW")

# Ubah nilai rpm
spek_motor['rpm'] = 1500

# Cetak ulang seluruh dictionary setelah perubahan
print("\nSpesifikasi motor setelah perubahan RPM:")
print(spek_motor)
```

```python
pembangkit = [
    {"nama": "Unit-1", "kapasitas_MW": 100, "status": "operasi"},
    {"nama": "Unit-2", "kapasitas_MW": 150, "status": "operasi"},
    {"nama": "Unit-3", "kapasitas_MW": 80,  "status": "pemeliharaan"},
]

print("Data pembangkit awal:")
for unit in pembangkit:
    print(unit)

# (a) Cetak nama unit kedua (indeks 1)
print(f"\nNama unit kedua: {pembangkit[1]['nama']}")

# (b) Ubah status salah satu unit (contoh: Unit-1 menjadi 'non-operasi')
pembangkit[0]['status'] = 'non-operasi'
print("\nData pembangkit setelah perubahan status Unit-1:")
for unit in pembangkit:
    print(unit)

# (c) Tambahkan satu unit baru ke list dengan append()
unit_baru = {"nama": "Unit-4", "kapasitas_MW": 120, "status": "persiapan"}
pembangkit.append(unit_baru)

# (d) Cetak seluruh data setelah penambahan unit baru
print("\nData pembangkit setelah penambahan Unit-4:")
for unit in pembangkit:
    print(unit)
```
